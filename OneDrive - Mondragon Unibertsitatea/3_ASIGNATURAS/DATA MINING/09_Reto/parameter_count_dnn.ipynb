# Number of parameters of DNNs

```python
import torch.nn as nn
```

```python
net1 = nn.Sequential(
    nn.Linear(10, 32),
    nn.ReLU(),
    nn.Linear(32, 1)
)
```

```python
net2 = nn.Sequential(
    nn.Linear(64, 128),
    nn.BatchNorm1d(128),
    nn.PReLU(num_parameters=1),
    nn.Linear(128, 64),
    nn.BatchNorm1d(64),
    nn.PReLU(num_parameters=1)
)
```

```python
net3 = nn.Sequential(
    nn.Linear(100, 256, bias=False),
    nn.BatchNorm1d(256),
    nn.PReLU(num_parameters=256),
    nn.Linear(256, 128),
    nn.PReLU(num_parameters=128),
    nn.Linear(128, 10),
    nn.Sigmoid()
)
```

```python
net4 = nn.Sequential(
    nn.Linear(20, 50),
    nn.BatchNorm1d(50),
    nn.LeakyReLU(negative_slope=0.1),
    nn.Linear(50, 50, bias=False),
    nn.BatchNorm1d(50),
    nn.PReLU(num_parameters=50),
    nn.Linear(50, 5)
)
```
