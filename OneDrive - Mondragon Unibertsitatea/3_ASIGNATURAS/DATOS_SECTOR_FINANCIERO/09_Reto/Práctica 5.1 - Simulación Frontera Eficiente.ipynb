```python
# Valores de cartera: Función.

def cartera(precios, pesos):

  # Calcular las rentabilidades diarias de la cartera
  rent_activos = precios.pct_change().dropna()
  rent_cartera = rent_activos.dot(pesos)

  # Calcular la matriz de covarianza anualizada
  covarianza = rent_activos.cov() * 252

  # Calcular la volatilidad anualizada de la cartera
  volatilidad_portfolio = np.sqrt(np.transpose(pesos) @ covarianza @ pesos)

  # Calcular la rentabilidad acumulada
  rtb_acu = (1+rent_cartera).cumprod()

  # Calcular la rentabilidad anualizada
  r = ((rtb_acu[-1]-rtb_acu[0]) / rtb_acu[0])
  rent_anualizada = ((1+r)**(252/len(precios)))-1

  return volatilidad_portfolio, rent_anualizada, rent_cartera, rtb_acu
```

```python
def rtb_anualizada(tipo, n, p0, p1, dias_habiles):

  if tipo == 'd':
    r=(p1-p0)/p0
    return ((1+r)**(dias_habiles/n))-1

  if tipo == 'm':
    r=(p1-p0)/p0
    return ((1+r)**(12/n))-1

  if tipo == 'y':
    r=(p1-p0)/p0
    return ((1+r)**(1/n))-1

  else:
    print("Por favor, introduzca 'd', 'm' o 'y' en el tipo")
    return
```
