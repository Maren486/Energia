```python
from datetime import datetime, timedelta

def cargarDatos(tickers, dias):
    fecha_fin = datetime.today()
    fecha_inicio = fecha_fin - timedelta(days=dias)
    
    datos = yf.download(
        tickers,
        start=fecha_inicio,
        end=fecha_fin
    )
    
    return datos
```

```python
# Función para calcular rentabilidades anualizadas

def rtb_anualizada(tipo, n, p0, p1, dias_habiles):  # n es cuanto tenemos (5 años), y dependerá de si tenemos días, meses o años
  if tipo == 'd':
    r = (p1-p0)/p0
    return ((1+r)**(dias_habiles/n))-1

  if tipo == 'm':
    r = (p1-p0)/p0
    return ((1+r)**(12/n))-1

  if tipo == 'y':  # si tenemos rentabilidades en términos de años (rentabilidad a 5 años, rent activa)
    r = (p1-p0)/p0
    return ((1+r)**(1/n))-1

  else:
    print("Por favor, introduzca 'd', 'm' o 'y' en el tipo")
    return
```

# Rentabilidad y volatilidad de los activos.

```python
rentabilidad_anualizada = rtb_anualizada('y', 5, p0, p1, 252)
rentabilidad_anualizada
```

**Volatilidad**

```python
rentabilidades_diarias = datos['Close'].pct_change().dropna()
rentabilidades_diarias.plot(figsize=(16,8))
plt.show()
```

```python
cuasivarianza = rentabilidades_diarias.var()  # por defecto pilla la ddof = 1
varianza = rentabilidades_diarias.var(ddof=0)
cuasidesviacion = rentabilidades_diarias.std()
desviacion = rentabilidades_diarias.std(ddof=0)
volatilidad_anualizada = desviacion * np.sqrt(252)  # 252 días de cotización en un año
volatilidad_anualizada
```

**Rentabilidad anualizada:** $r_{anualizada} = (1 + r_{activa})^n-1$

**Rentabilidad de la cartera:** $R\cdot X^T$

**Covarianza anualizada:** $M_{cov-anualizada} = M_{cov} \cdot 252 $

**Volatilidad de la cartera:** $\sigma_{cartera} = \sqrt{X \cdot M_{cov-anualizada} \cdot X^T}$

**Rentabilidad acumulada:** $R_{acu} = (1+RC_1)\cdot(1+RC_2)\cdot ... \cdot (1+RC_m) $

```python
X = np.array([0.33, 0.33, 0.34])
```

```python
# Volatilidad cartera: función.
def volatilidad_cartera(X, M_cov_anualizada):
    volatilidad = np.sqrt(
        X @ M_cov_anualizada @ X.T
    )
    return volatilidad
```

```python
# Rentabilidad y volatilidad anualizadas de la cartera.
rentabilidad_cartera = rentabilidad_anualizada @ rentabilidad_anualizada.T
rentabilidad_cartera
```

```python
# Rentabilidad acumulada de la cartera.

# Rentabilidad diaria de la cartera
rentabilidad_cartera_diaria = rentabilidades_diarias @ X.T

# Rentabilidad acumulada de la cartera
rentabilidad_acumulada = (1 + rentabilidad_cartera_diaria).prod() - 1

print(rentabilidad_acumulada)
```

```python
#VOLATILIDAD DE LA CARTERA
M_cov = rentabilidades_diarias.cov()
# Matriz de covarianzas anualizada
M_cov_anualizada = M_cov * 252
```

```python
volatilidad_cartera = volatilidad_cartera(X, M_cov_anualizada)
volatilidad_cartera
```
